```python
import random

# -------------------------------------------------
# Part A – Machine Initialization
# -------------------------------------------------

machines = ['A', 'B', 'C']

# Number of machines
M = 3

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# -------------------------------------------------
# Part B – Initialize N, R and Q tables
# -------------------------------------------------

# N = number of times each machine is selected
N = {machine: 0 for machine in machines}

# R = total reward received from each machine
R = {machine: 0 for machine in machines}

# Q = estimated value of each machine
Q = {machine: 0.0 for machine in machines}

# Store total reward received
total_reward = 0

print("==============================================")
print("   ε-GREEDY MULTI-ARMED BANDIT")
print("==============================================")

# -------------------------------------------------
# Part C – ε-Greedy Decision
# -------------------------------------------------

for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    if r < epsilon:
        # Exploration: randomly select a machine
        selected_machine = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation: select machine with highest Q-value
        max_q = max(Q.values())

        # If multiple machines have the same Q-value,
        # randomly select one among them
        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    # -------------------------------------------------
    # Part D – Reward Collection
    # -------------------------------------------------

    print(f"\nPlay {play}")
    print(f"Random number = {r:.4f}")
    print(f"Decision = {decision}")
    print(f"Play with machine {selected_machine}")

    # Handle empty input for reward
    reward_input = input("Enter reward: ")
    if reward_input == '':
        reward = 0.0  # Default value for empty input
    else:
        reward = float(reward_input)

    # -------------------------------------------------
    # Part E – Q-Value Update
    # -------------------------------------------------

    # Increase participation count
    N[selected_machine] += 1

    # Add reward
    R[selected_machine] += reward

    # Incremental sample-average formula:
    # Q(a) = Q(a) + [R - Q(a)] / N(a)

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    # Add to total reward
    total_reward += reward

    print(f"Updated Q-value of machine {selected_machine} = "
          f"{Q[selected_machine]:.2f}")

# -------------------------------------------------
# Part F – Display Results
# -------------------------------------------------

print("\n==============================================")
print("             FINAL RESULTS")
print("==============================================")

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]:.2f}\t"
        f"{Q[machine]:.2f}"
    )

print("\n==============================================")
print(f"Total Reward after {N_play} plays = {total_reward:.2f}")
print("==============================================")
```
